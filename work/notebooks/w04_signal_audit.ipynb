## 1. Distributions

*Look before deciding: distributions of your key fields. Note the heavy tails.*

```python
import duckdb
from google.colab import userdata
from huggingface_hub import hf_hub_download

hf_token = userdata.get('HF_TOKEN')

file_path = "fact_content_daily_performance/month=2026-03/data_0.parquet"
local_parquet_path = hf_hub_download(
    repo_id="FlyRank/internship-warehouse",
    filename=file_path,
    repo_type="dataset",
    token=hf_token
)

con = duckdb.connect()

feature_df = con.execute(f"""
    SELECT
        content_hash_id,
        COALESCE(SUM(gsc_impressions), 0) AS total_gsc_impressions,
        COALESCE(SUM(gsc_clicks), 0) AS total_gsc_clicks,
        CASE WHEN SUM(gsc_impressions) > 0
             THEN SUM(gsc_clicks) * 1.0 / SUM(gsc_impressions)
             ELSE 0 END AS ctr,
        COALESCE(SUM(ga4_pageviews), 0) AS total_ga4_pageviews,
        COALESCE(SUM(ga4_sessions), 0) AS total_ga4_sessions,
        MAX(CASE WHEN client_has_gsc THEN 1 ELSE 0 END) AS has_gsc_bool,
        MAX(CASE WHEN client_has_ga4 THEN 1 ELSE 0 END) AS has_ga4_bool,
        COUNT(*) AS days_observed
    FROM '{local_parquet_path}'
    WHERE gsc_data_available IS TRUE
    GROUP BY content_hash_id
""").df()

feature_df = feature_df.fillna(0)

print(f"Feature vector shape: {feature_df.shape}")
display(feature_df.head())
```

```python
import matplotlib.pyplot as plt

# Reuse con, local_parquet_path, feature_df from earlier notebooks in this session
# If running fresh, re-run the setup + feature_df cells first (Runtime → Run all)

fig, axes = plt.subplots(2, 3, figsize=(15, 8))

fields = ['total_gsc_impressions', 'total_gsc_clicks', 'ctr',
          'total_ga4_pageviews', 'total_ga4_sessions', 'days_observed']

for ax, field in zip(axes.flat, fields):
    ax.hist(feature_df[field], bins=50)
    ax.set_title(field)
    ax.set_yscale('log')  # heavy tails make linear scale useless here

plt.tight_layout()
plt.show()

# Numeric summary alongside the plots
display(feature_df[fields].describe(percentiles=[.5, .9, .95, .99]))
```

All five volume-based fields (impressions, clicks, pageviews, sessions) are heavily right-skewed —
a small number of content items carry most of the traffic, and the median sits far below the mean.
`ctr` is bounded [0,1] and closer to a normal-ish shape but still has a spike at 0 (pages with zero clicks).
`days_observed` clusters near 31 (full-month coverage) with a smaller tail of partial-month rows.

## 2. Signal test #1 / #2 / #3 (verdict each)

*Three safe signals, each with a mini-test and a verdict: CONFIRMED / OPPOSITE / MIXED / FALSE.*

```python
import numpy as np

# --- Signal 1: Higher CTR associates with more clicks per impression already captured,
# so test whether higher ga4_sessions associates with higher ctr (engagement carries over)
median_sessions = feature_df['total_ga4_sessions'].median()
high_sessions = feature_df[feature_df['total_ga4_sessions'] > median_sessions]['ctr']
low_sessions = feature_df[feature_df['total_ga4_sessions'] <= median_sessions]['ctr']

print("Signal 1: Higher GA4 sessions -> higher CTR?")
print(f"  Mean CTR, high-session group: {high_sessions.mean():.4f}")
print(f"  Mean CTR, low-session group:  {low_sessions.mean():.4f}")
# Verdict written below based on printed values
```

```python
# --- Signal 2: More days observed (fuller month coverage) associates with higher impression volume
median_days = feature_df['days_observed'].median()
full_month = feature_df[feature_df['days_observed'] >= median_days]['total_gsc_impressions']
partial_month = feature_df[feature_df['days_observed'] < median_days]['total_gsc_impressions']

print("Signal 2: Full-month coverage -> higher impressions?")
print(f"  Mean impressions, full-month group:    {full_month.mean():.2f}")
print(f"  Mean impressions, partial-month group: {partial_month.mean():.2f}")
```

```python
# --- Signal 3: has_ga4_bool associates with higher total_gsc_clicks (cross-platform instrumentation
# correlates with overall content investment/quality)
ga4_yes = feature_df[feature_df['has_ga4_bool'] == 1]['total_gsc_clicks']
ga4_no = feature_df[feature_df['has_ga4_bool'] == 0]['total_gsc_clicks']

print("Signal 3: GA4 instrumented -> higher GSC clicks?")
print(f"  Mean clicks, GA4-instrumented: {ga4_yes.mean():.2f}")
print(f"  Mean clicks, not instrumented: {ga4_no.mean():.2f}")
```

## 3. The flag-linked test

*Pick a signal one of FlyRank's real flags relies on. Does the data support the rule's assumption?*

```python
# Flag tested: ML-04's decay flag — position worse than panel median (pos_last30 > median)
# Assumption behind the rule: pages with worse-than-median position also show weaker engagement
# (lower CTR, fewer clicks) — i.e. position decay is a real engagement-decay signal, not noise.

# Recompute avg position separately — NOT added to feature_df, used only to build the flag
position_lookup = con.execute(f"""
    SELECT
        content_hash_id,
        AVG(gsc_avg_position) AS avg_position
    FROM '{local_parquet_path}'
    WHERE gsc_data_available IS TRUE
    GROUP BY content_hash_id
""").df()

median_position = position_lookup['avg_position'].median()
position_lookup['flagged_decay'] = position_lookup['avg_position'] > median_position

flag_test = feature_df.merge(position_lookup[['content_hash_id', 'flagged_decay']], on='content_hash_id')

flagged = flag_test[flag_test['flagged_decay'] == True]
not_flagged = flag_test[flag_test['flagged_decay'] == False]

print(f"Rows flagged (position > median): {len(flagged)} ({len(flagged)/len(flag_test)*100:.1f}%)")
print(f"\nMean CTR, flagged:     {flagged['ctr'].mean():.4f}")
print(f"Mean CTR, not flagged: {not_flagged['ctr'].mean():.4f}")
print(f"\nMean GSC clicks, flagged:     {flagged['total_gsc_clicks'].mean():.2f}")
print(f"Mean GSC clicks, not flagged: {not_flagged['total_gsc_clicks'].mean():.2f}")
print(f"\nMean GA4 sessions, flagged:     {flagged['total_ga4_sessions'].mean():.2f}")
print(f"Mean GA4 sessions, not flagged: {not_flagged['total_ga4_sessions'].mean():.2f}")
```

**Flag tested:** ML-04's decay flag — `pos_last30 > 81.7` (position worse than panel median)
**Assumption behind it:** worse-than-median position is a real signal of engagement decay, not just noise in the ranking.

**What I found:** Flagged content (50.0% of rows, 88,369 pages) showed lower mean CTR (0.0026 vs 0.0066 —
roughly 2.5x lower) and lower mean GSC clicks (2.43 vs 6.87 — roughly 2.8x lower) than non-flagged content,
consistent with the flag's assumption. However, mean GA4 sessions ran in the opposite direction — flagged
content actually showed *higher* average sessions (9.64 vs 4.38) than non-flagged content.

**Verdict:** MIXED. Search-side engagement (CTR, clicks) supports the flag's assumption — worse position
does associate with weaker search engagement. But on-site engagement (GA4 sessions) moves the opposite way,
so "position decay = overall engagement decay" doesn't hold across both data sources at once.

## 4. What this means in practice

*Two or three sentences: what a content team should take from this.*

The position-decay flag is a reasonable search-engagement signal on its own — CTR and clicks both drop
meaningfully for flagged pages, so it's doing what it claims to do on the search side. But the GA4 session
pattern flips the story: flagged pages aren't seeing an overall engagement collapse, they're likely getting
traffic through other channels (direct, referral, social) even as search performance weakens. A content team
should read this flag narrowly, as "search visibility is declining," not broadly as "this content is dying" —
and should check on-site engagement separately before deprioritizing or rewriting flagged pages.
